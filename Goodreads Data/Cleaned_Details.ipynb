```python
import pandas as pd

def clean_data(df):
    # Replace all instances of "Not specified" with "Adult" in column: 'Audience Genre'
    df['Audience Genre'] = df['Audience Genre'].str.replace("Not specified", "Adult", case=False, regex=False)
    return df

# Loaded variable 'df' from URI: c:\Programming\Python\msba_capstone_dragonsteel\Goodreads Data\Combined_Details.csv
df = pd.read_csv(r'c:\Programming\Python\msba_capstone_dragonsteel\Goodreads Data\Combined_Details.csv')

df_clean = clean_data(df.copy())
df_clean.head()
```
